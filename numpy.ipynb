**NUMPY**

```python
import numpy as np
```

```python
#1d array
arr1=np.array([1,2,3,4,5])
print(arr1)
```

```python
# 2d array
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2)
```

```python
#zeros
print(np.zeros((2,3)))

#ones
print(np.ones((3,3)))
```

```python
# range

print(np.arange(0,10,2))
```

```python
#linspace

print(np.linspace(0,1,5))
```

```python
# array properties
"""
print(arr2.shape)      # shape
print(arr2.ndim)       # number of dimensions
print(arr2.size)       # total elements
print(arr2.dtype)      # data type

"""
```

```python
# indexing and slicing

arr = np.array([10,20,30,40,50])

print(arr[0])
print(arr[-1])
print(arr[1:4])      #1 is inclusive and 4 is exclusive
```

```python
# 2D
arr = np.array([[1,2,3],
                [4,5,6]])

print(arr[0,1])
print(arr[:,1])
```

```python
# mathematical operations

a= np.array([1,2,3])
b= np.array([1,2,3])

print(a+b)

print(a-b)
print(a*b)
print(a/b)

print(np.sqrt(a))
print(np.sum(a))
print(np.mean(a))
print(np.max(a))
```

```python
# reshaping

arr=np.arange(1,10)
reshaped=arr.reshape(3,3)
print(reshaped)
```
